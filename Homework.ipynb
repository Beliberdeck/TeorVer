### Задача 1.10

### Задача 1.48
> В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения 11 очков; б) выигрыша.

```python
m = 0
n = 0

for i in range(1, 7):
    for j in range(1, 7):
        for k in range(1, 7):
            m += 1
            if i + j + k == 11:
                n += 1

round(n / m, 4)
```

### Задача 1.50
> Два лица условились встретиться в определенном месте между 18 и 19 ч и договорились, что пришедший первым ждет другого в течение 15 мин., после чего уходит. Найти вероятность их встречи, если приход каждого в течение указанного часа может произойти в любое время и моменты прихода независимы.

```python
from scipy.integrate import quad

def f1(x):
    return min(x + 15/60, 1)

def f2(x):
    return max(x - 15/60, 0)

r1 = quad(f1, 0, 1)
r2 = quad(f2, 0, 1)

round(r1[0] - r2[0], 4)
```

* 1) $P_9(4) = C_9^4 \cdot 0,2^4 \cdot 0,8^5$

```python
round(stats.binom.cdf(2, 9, 0.2), 4)
```

```python
round(1 - stats.binom.cdf(1, 9, 0.2), 4)
```

```python
round(stats.binom.pmf(1, 9, 0.2) + stats.binom.pmf(2, 9, 0.2), 4)
```

### Задача 2.11
> По результатам проверок налоговыми инспекциями установлено, что в среднем каждое второе малое предприятие региона имеет нарушение финансовой дисциплины. Найти вероятность того, что из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины: а) 480 предприятий; б) наивероятнейшее число предприятий; в) не менее 480; г) от 480 до 520.


* а) $n = 1000, p = 0,5, q = 0,5, npq = 250$
$x = \frac{480 - 500}{\sqrt{250}} \approx -1,265$
$P_{1000}(480) \approx \frac{\varphi(x)}{\sqrt{npq}}$


```python
n = 1000
p = 0.5
s = math.sqrt(n * p * (1 - p))
x = (480 - n * p) / s
round(stats.norm.pdf(x) / s, 4)
```

* б) $np - q \le m_0 \le np + p \implies m_0 = 500$


```python
x = (500 - n * p) / s
round(stats.norm.pdf(x) / s, 4)
```

* в) $P(480 \le m \le 1000) = \Phi(x_2) - \Phi(x_1)$


```python
x1 = (480 - n * p) / s
x2 = (1000 - n * p) / s
round(stats.norm.cdf(x2) - stats.norm.cdf(x1), 4)
```

* г) $P(480 \le m \le 520) = \Phi(x_2) - \Phi(x_1)$


```python
x1 = (480 - n * p) / s
x2 = (520 - n * p) / s
round(stats.norm.cdf(x2) - stats.norm.cdf(x1), 4)
```
