Парадокс ящиков Бертрана.

У Марсика есть три коробки с игрушечными болидами: в первой два болида Red Bull, во второй два болида Ferrari, в третьей один Red Bull и один Ferrari. Марсик наугад выбирает коробку и достаёт из неё лапой случайный болид. Это оказывается болид Red Bull.

Какова вероятность, что второй болид в этой же коробке тоже Red Bull? Смоделировать решение.

```python
import matplotlib.pyplot as plt
import random
```

## Симуляция

```python
N = 100000

boxes = [
    ['RB', 'RB'],
    ['F', 'F'],
    ['RB', 'F'],
]

c_redbull = 0
c_both = 0    

n_list = []
freq = []

for i in range(N):
    box = random.choice(boxes)

    first = random.randint(0, 1)
    second = 1 - first

    if box[first] == 'RB':
        c_redbull += 1

        if box[second] == 'RB':
            c_both += 1

        n_list.append(c_redbull)
        freq.append(c_both / c_redbull)

res = c_both / c_redbull
res_true = 2 / 3

print("Опыт:", res)
print("Теоретическая вероятность:", res_true)
```

## Визуализация

```python
plt.figure(figsize=(10, 5))

plt.plot(n_list, freq, color='blue', label='Частота')
plt.axhline(2 / 3, color='red', linestyle='--', label='P = 2/3')
plt.axhline(1 / 2, color='gray', linestyle='--', label='Интуитивный ответ 1/2')

plt.xscale('log')
plt.ylim(0, 1)

plt.title("Марсик достал Red Bull: второй болид тоже Red Bull?")
plt.xlabel("Число опытов, где первый болид - Red Bull")
plt.ylabel("Частота")
plt.legend()

plt.show()
```
