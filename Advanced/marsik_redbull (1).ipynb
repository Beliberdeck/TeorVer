Мейн-кун Марсик сидит у мешка с 10 игрушечными болидами: 3 болида Red Bull и 7 болидов других команд. Он наугад достаёт один болид, смотрит и кладёт обратно.

Какова вероятность, что Марсик достанет болид Red Bull? Показать, что частота стремится к классической вероятности при увеличении числа опытов.

```python
import matplotlib.pyplot as plt
import random
```

## Симуляция

```python
N = 5000

bag = [1] * 3 + [0] * 7  # 1 - Red Bull, 0 - другая команда

c = 0
n_list = []
freq = []

for i in range(1, N + 1):
    if random.choice(bag) == 1:
        c += 1
    n_list.append(i)
    freq.append(c / i)

res = c / N
res_true = 3 / 10

print("Опыт:", res)
print("Теоретическая вероятность:", res_true)
```

## Визуализация

```python
plt.figure(figsize=(10, 5))

plt.plot(n_list, freq, color='blue', label='Частота')
plt.axhline(res_true, color='red', linestyle='--', label='P = 3/10')

plt.xscale('log')
plt.ylim(0, 1)

plt.title("Марсик достаёт болид Red Bull")
plt.xlabel("Число опытов")
plt.ylabel("Частота")
plt.legend()

plt.show()
```
